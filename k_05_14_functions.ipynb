```python
def add_numbers(a: int, b: int) -> int:
    """
    Add two integers and return the result.

    Parameters:
    a (int): First integer
    b (int): Second integer

    Returns:
    int: Sum of a and b
    """
    return a + b
```

```python
def is_even(num):
    """
    This function returns if given number is odd or even
    input- any valid integer
    output-even / odd
    """
    if num%2==0:
        return'even'
    else:
        return'odd'

for i in range (1,11):
    x=is_even(i)
    print(x)
```
