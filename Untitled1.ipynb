```python
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature


df = pd.read_csv("/content/andaman_nicobar_M4_catalog.csv")
fig = plt.figure(figsize=(12,8))
ax = plt.axes(projection=ccrs.PlateCarree())

# Add coastlines and borders
ax.coastlines(alpha=0.9)
ax.add_feature(cfeature.BORDERS, linestyle=':')
ax.add_feature(cfeature.LAND, color='lightgrey')


# Plot epicenters colored by depth
sc = ax.scatter(df['longitude'],df['latitude'],c=df['depth_km'],cmap='viridis',s=20,transform=ccrs.PlateCarree())

# Colorbar
cbar = plt.colorbar(sc, ax=ax, shrink=0.7)
cbar.set_label('Depth (km)')

# Gridlines
gl = ax.gridlines(draw_labels=True)
gl.top_labels = False
gl.right_labels = False

plt.title("Earthquake Epicenters Colored by Depth")
plt.show()
```

**CATALOG STATISTICS**

```python
import pandas as pd

df = pd.read_csv("/content/andaman_nicobar_M4_catalog.csv")

print(df.head())
```

```python
df = df.dropna(subset=["magnitude"])
mags = df["magnitude"].values
len(mags)  #so no missing values
```

```python
import matplotlib.pyplot as plt
import numpy as np

bins = np.arange( mags.min(), mags.max(), 0.1)

plt.hist(mags,bins=bins,edgecolor='black')

plt.xlabel("Magnitude")
plt.ylabel("Frequency")
plt.title("Magnitude Frequency Distribution")
plt.show()
```

```python
hist, edges = np.histogram(mags,bins=bins)
Mc = edges[np.argmax(hist)]

print("Mc =", round(Mc,2))
```

```python
complete_mags = mags[mags >= Mc]

print("Number of complete events:",len(complete_mags))
```

```python
import numpy as np
dM = 0.1
mean_mag = np.mean(complete_mags)
b = 1 / (np.log(10) * (mean_mag - Mc + dM/2))
print("b-value =", round(b,3))
print("b_mean=",b.mean())
```

```python
N = len(complete_mags)
sigma_b = b / np.sqrt(N)
ci_low=np.percentile(b,2.5)
ci_high=np.percentile(b,97.5)
print(f"b = {b:.3f} ± {sigma_b:.3f}")
print(f"95% CI: [{ci_low:.3f}, {ci_high:.3f}]")
```

```python
bins = np.arange(Mc,complete_mags.max(),0.1)
hist, edges = np.histogram(complete_mags,bins=bins)
cum_counts = np.cumsum(hist[::-1])[::-1]

plt.figure(figsize=(7,5))

plt.semilogy(edges[:-1],cum_counts,'+')
plt.xlabel("Magnitude")
plt.ylabel("Cumulative Count")
plt.title("Gutenberg-Richter Plot")
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import numpy as np

from openquake.hmtk.seismicity.catalogue import Catalogue

from openquake.hmtk.seismicity.declusterer.dec_gardner_knopoff import (GardnerKnopoffType1)

from openquake.hmtk.seismicity.declusterer.distance_time_windows import (GardnerKnopoffWindow)
```

```python
df=pd.read_csv("/content/andaman_nicobar_M4_catalog.csv")
df["time"]=pd.to_datetime(df["time"])
print(df.head())
```

```python
cat = Catalogue()

cat.data = {
    "year": df["time"].dt.year.values,
    "month": df["time"].dt.month.values,
    "day": df["time"].dt.day.values,

    "latitude": df["latitude"].values,
    "longitude": df["longitude"].values,
    "depth_km": df["depth_km"].values,

    "magnitude": df["magnitude"].values}

cat.number_earthquakes = len(df)
config = {
    "time_distance_window": GardnerKnopoffWindow(),
    "fs_time_prop": 1.0}
```

```python
gk = GardnerKnopoffType1()

vcl, flagvector = gk.decluster( cat, config)
declustered = df[flagvector == 0].copy()
```

```python
print("Original:", len(df))
print("Declustered:", len(declustered))
```
