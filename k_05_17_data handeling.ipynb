```python
import pickle # Move import to the top level

#pickling and unpickling
class person :
  def __init__ (self,name,age):
    self.name=name
    self.age=age

  def display_info(self):
    print('I am',self.name,'and I am',self.age,'years old')


p=person('kunmun',23)

#pickling
import pickle
with open ('person.pkl','wb')as f:
  pickle.dump(p,f) 

#unpickling
with open ('person.pkl','rb')as f:
  p=pickle.load(f)
  p.display_info()
```

```python
import json

class person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

def show_object(obj):

        return {'name': obj.name, 'age': obj.age}

    
p_instance = person('kunmun', 23)

with open('data json','w') as f:

    json.dump(p_instance, f, default=show_object, indent=4)
```
