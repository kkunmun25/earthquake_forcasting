```python
from dataclasses import dataclass
from datetime import datetime


# Parent class
@dataclass
class Earthquake:
    
    mag: float
    depth: float
    lat: float
    lon: float
    time: datetime

    def info(self):
        print(f"Magnitude: {self.mag}")
        print(f"Depth: {self.depth} km")
        print(f"Latitude: {self.lat}")
        print(f"Longitude: {self.lon}")
        print(f"Time: {self.time}")


# Child class: Local Earthquake
@dataclass
class LocalEarthquake(Earthquake):
    
    

    def category(self):
        return "Local Earthquake"


# Child class: Teleseismic Earthquake
@dataclass
class TeleseismicEarthquake(Earthquake):
    
    

    def category(self):
        return "Teleseismic Earthquake"




local_eq = LocalEarthquake(mag=4.5,depth=12,lat=35.6,lon=140.1,time=datetime.now())

tele_eq = TeleseismicEarthquake(mag=7.2,depth=50,lat=-10.2,lon=120.5,time=datetime.now())

# Using methods
local_eq.info()
print(local_eq.category())

print()

tele_eq.info()
print(tele_eq.category())
```
