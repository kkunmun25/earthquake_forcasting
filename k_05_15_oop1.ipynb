```python
from math import radians, sin, cos, sqrt, atan2
from datetime import datetime

class Earthquake:

    def __init__(self, mag, depth, lat, lon, time):
        self.mag = mag
        self.depth = depth
        self.lat = lat
        self.lon = lon
        self.time = time

    def distance_to(self, other):
        """
        Calculate great-circle distance between two earthquakes
        using the Haversine formula.

        Returns distance in kilometers.
        """
        R = 6371  # Earth radius in km

        lat1 = radians(self.lat)
        lon1 = radians(self.lon)

        lat2 = radians(other.lat)
        lon2 = radians(other.lon)

        dlat = lat2 - lat1
        dlon = lon2 - lon1
        a=  (sin(dlat / 2) ** 2
            + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2)


        c = 2 * atan2(sqrt(a), sqrt(1 - a))

        return R * c
    
    @classmethod
    def from_csv_row(cls, row):
        """
        Create Earthquake object from CSV row.

        Expected format:
        [mag, depth, lat, lon, time]
        """

        mag = float(row[0])
        depth = float(row[1])
        lat = float(row[2])
        lon = float(row[3])

        time = datetime.fromisoformat(row[4])

        return cls(mag, depth, lat, lon, time)

eq1 = Earthquake (mag=6.5,
    depth=10,
    lat=35.6895,
    lon=139.6917,
    time=datetime.now())

row = ["5.8", "20", "34.0522", "-118.2437", "2026-05-14T10:30:00"]

eq2 = Earthquake.from_csv_row(row)

print("Magnitude:", eq2.mag)

distance = eq1.distance_to(eq2)

print(f"Distance: {distance:.2f} km")
```
