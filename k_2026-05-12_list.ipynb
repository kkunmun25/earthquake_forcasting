```python
#adding 1 to 10 number in a list
L=[i for i in range(1,11)]
print(L)
```

```python
#scalar multiplication on vector
v=[2,3,4]
s=-2
L=[s*i for i in v]
print(L)
```

```python
#adding square
l=[1,2,3]
[i*2 for i in l]
```

```python
#print all numbers divisible by 5 in range 1 to 51
[i for i in range(1,51) if i%5==0]
```

```python
#nested if with list comprehation
#add new list from my_fruits n items if fruits exit in basket and also started with'a'
basket=['apple','orange','banana','kiwi']
my_fruits=['apple','kiwi','grapes']
[i for i in  my_fruits if i in basket if i.startswith('a')]
```

```python
#print a (3,3) matrix using list comprehation
[[i*j for i in range(1,4)] for j in range(1,4)]
```

```python
#list comprehation in two lists together-cartesian product\
L1=[1,2,3]
L2=[4,5,6]
[i*j for i in L1 for j in L2]
```

```python
#Write a list comprehension that can flatten a nested list
matrix = [[1,2,3],[4,5,6],[7,8,9]]
l=[j for i in matrix for j in i]
print(l)
```

```python
#Write a list comprehension to print the following matrix
L=[[0, 1, 2], [3, 4, 5], [6, 7, 8]]
[i for i in L]
```

```python
#Write a list comprehension that can transpose a given matrix

matrix = [[1,2,3],[4,5,6],[7,8,9]]
transposed_matrix = [[row[i] for row in matrix] for i in range(len(matrix[0]))]
print(transposed_matrix)
```
